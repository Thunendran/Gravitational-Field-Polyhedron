```python
#!/usr/bin/env python3
import numpy as np
import os

try:
    ROOT = os.path.dirname(os.path.dirname(os.path.abspath(__file__)))
except NameError:
    # Running in notebook or interactive session
    ROOT = os.path.abspath(os.path.join(os.getcwd(), "Polyhedron"))

DATA_DIR = os.path.join(ROOT, "data")
os.makedirs(DATA_DIR, exist_ok=True)
```

```python
# ---------- Icosahedron (unit scale) ----------
phi = (1.0 + np.sqrt(5.0)) / 2.0

# 12 vertices (classic coordinates)
verts = np.array([
    [-1,  phi,  0],
    [ 1,  phi,  0],
    [-1, -phi,  0],
    [ 1, -phi,  0],
    [ 0, -1,  phi],
    [ 0,  1,  phi],
    [ 0, -1, -phi],
    [ 0,  1, -phi],
    [ phi, 0, -1],
    [ phi, 0,  1],
    [-phi, 0, -1],
    [-phi, 0,  1],
], dtype=np.float64)

# Normalize to unit circumradius (optional but nice)
r = np.linalg.norm(verts, axis=1).max()
verts = verts / r

# 20 faces (1-based for Julia/Matlab; Python can adjust as needed)
faces = np.array([
    [1, 12,  6],
    [1,  6,  2],
    [1,  2,  8],
    [1,  8, 11],
    [1, 11, 12],
    [2,  6, 10],
    [6, 12,  5],
    [12, 11,  3],
    [11,  8,  7],
    [8,  2,  9],
    [4, 10,  5],
    [4,  5,  3],
    [4,  3,  7],
    [4,  7,  9],
    [4,  9, 10],
    [5, 10,  6],
    [3,  5, 12],
    [7,  3, 11],
    [9,  7,  8],
    [10,  9,  2],
], dtype=np.int32)

# ---------- Evaluation points ----------
# Include all vertices (12) + 100,000 random points (in a reasonable box)
np.random.seed(12345)
N_rand = 99_988
rand_pts = np.random.uniform(-1.5, 1.5, size=(N_rand, 3))
points = np.vstack([verts, rand_pts])  # total: 100,000

# ---------- Save CSVs ----------
np.savetxt(os.path.join(DATA_DIR, "icosahedron_vertices.csv"), verts, delimiter=",")
np.savetxt(os.path.join(DATA_DIR, "icosahedron_faces.csv"), faces, fmt="%d", delimiter=",")
np.savetxt(os.path.join(DATA_DIR, "eval_points_100k_plus_vertices.csv"), points, delimiter=",")

print("Saved:")
print("  data/icosahedron_vertices.csv")
print("  data/icosahedron_faces.csv")
print("  data/eval_points_100k_plus_vertices.csv (shape:", points.shape, ")")
```
